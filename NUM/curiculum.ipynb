```python
import pandas as pd

# Load the datasets
df_2017 = pd.read_csv('-2017-autumn.csv')
df_2023 = pd.read_csv('-2023-autumn1.csv')

# Print out their sizes
print(f"2017 Dataset Shape: {df_2017.shape}")
print(f"2023 Dataset Shape: {df_2023.shape}")

# Inspect the columns
print("\n2017 Columns:", df_2017.columns.tolist())
print("2023 Columns:", df_2023.columns.tolist())
```

```python
# Extract the features we need
data_2017 = df_2017[['Монгол_нэр', 'Харьяалах_бүтэц']].copy()
data_2023 = df_2023[['Монгол_нэр', 'Харьяалах_бүтэц']].copy()

# Combine them on top of each other
combined_df = pd.concat([data_2017, data_2023], axis=0)
print(f"Size BEFORE removing duplicates: {combined_df.shape}")

# Drop rows where the course name is exactly the same
cleaned_df = combined_df.drop_duplicates(subset=['Монгол_нэр'])
print(f"Size AFTER removing duplicates: {cleaned_df.shape}")

# Drop any blank rows if they exist
cleaned_df = cleaned_df.dropna()
```

```python
print("Unique Schools in your cleaned dataset:")
for school in cleaned_df['Харьяалах_бүтэц'].unique():
    print(f"- {school}")
```

```python
# Create a dictionary to map legacy/alternative names to modern unified names
school_mapping = {
    'Олон улсын харилцаа, нийтийн удирдлагын сургууль': 'Улс төр судлал, олон улсын харилцаа, нийтийн удирдлагын сургууль',
    'Хэрэглээний шинжлэх ухаан, инженерчлэлийн сургууль': 'Мэдээллийн технологи, электроникийн сургууль', # Mapping to the dominant tech branch for simplicity
    'Монгол Улсын Их Сургууль': 'Бусад',
    'Элсэлт, бүртгэлийн хэлтэс': 'Бусад',
    'Бүртгэл, дүн шинжилгээний төв': 'Бусад',
    'Шинжлэх ухаан технологийн Улаанбаатар парк': 'Бусад'
}

# Apply the mapping. If a school isn't in the dictionary, keep its original name
cleaned_df['Харьяалах_бүтэц'] = cleaned_df['Харьяалах_бүтэц'].replace(school_mapping)

# Print out the newly unified distribution
print(cleaned_df['Харьяалах_бүтэц'].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# 1. Transform text labels into numeric targets (e.g., 0, 1, 2...)
label_encoder = LabelEncoder()
y = label_encoder.fit_transform(cleaned_df['Харьяалах_бүтэц'])

# 2. Vectorize Mongolian text using word fragments (n-grams) to catch root words
vectorizer = TfidfVectorizer(ngram_range=(1, 2), min_df=2)
X = vectorizer.fit_transform(cleaned_df['Монгол_нэр'])

# 3. Stratified split to ensure equal proportion of classes in train and test sets
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# 4. Initialize the model with balanced weights to handle smaller departments
model = LogisticRegression(max_iter=1000, class_weight='balanced')
model.fit(X_train, y_train)

# 5. Evaluate the results
y_pred = model.predict(X_test)
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))
```

```python
# --- STEP 7: HIGH-SPEED OPTIMIZED MODEL FOR MONGOLIAN TEXT ---

# 1. Vectorize using constrained character chunks to keep it fast
vectorizer_opt = TfidfVectorizer(
    analyzer='char_wb', 
    ngram_range=(3, 4),   # 3 to 4 characters captures roots perfectly
    min_df=5,             # Ignore ultra-rare fragments
    max_features=15000    # Cap the matrix size to save memory
)
X_opt = vectorizer_opt.fit_transform(cleaned_df['Монгол_нэр'])

# 2. Re-split the features
X_train_o, X_test_o, y_train_o, y_test_o = train_test_split(
    X_opt, y, test_size=0.2, stratify=y, random_state=42
)

# 3. Train using n_jobs=-1 to unlock parallel multi-core computing speed
model_opt = LogisticRegression(max_iter=500, class_weight='balanced', C=2.0, n_jobs=-1)
model_opt.fit(X_train_o, y_train_o)

# 4. Check the new results quietly (ignoring zero division warnings)
y_pred_o = model_opt.predict(X_test_o)
print(classification_report(y_test_o, y_pred_o, target_names=label_encoder.classes_, zero_division=0))
```

```python
# --- STEP 8: THE MAX ACCURACY PIPELINE ---

# 1. Filter out 'Бусад' and 'Нийгмийн ухааны салбар' to eliminate noise
clean_academic_df = cleaned_df[~cleaned_df['Харьяалах_бүтэц'].isin(['Бусад', 'Нийгмийн ухааны салбар'])].copy()

# 2. Re-encode targets
label_encoder_final = LabelEncoder()
y_final = label_encoder_final.fit_transform(clean_academic_df['Харьяалах_бүтэц'])

# 3. Vectorize text using tight character chunks
vectorizer_final = TfidfVectorizer(analyzer='char_wb', ngram_range=(3, 4), min_df=3, max_features=20000)
X_final = vectorizer_final.fit_transform(clean_academic_df['Монгол_нэр'])

# 4. Re-split
X_train_f, X_test_f, y_train_f, y_test_f = train_test_split(
    X_final, y_final, test_size=0.2, stratify=y_final, random_state=42
)

# 5. Train with high complexity (C=10.0)
model_final = LogisticRegression(max_iter=1000, class_weight='balanced', C=10.0, n_jobs=-1)
model_final.fit(X_train_f, y_train_f)

# 6. Print final results
y_pred_f = model_final.predict(X_test_f)
print(classification_report(y_test_f, y_pred_f, target_names=label_encoder_final.classes_, zero_division=0))
```

```python
# --- STEP 9: THE NUCLEAR OPTION (RANDOM FOREST) ---
from sklearn.ensemble import RandomForestClassifier

# 1. We use the exact same clean data from Step 8
# 2. We initialize a Random Forest with 150 parallel trees
rf_model = RandomForestClassifier(n_estimators=150, class_weight='balanced', random_state=42, n_jobs=-1)

# 3. Train the trees
rf_model.fit(X_train_f, y_train_f)

# 4. Check if it beats 62%
y_pred_rf = rf_model.predict(X_test_f)
print(classification_report(y_test_f, y_pred_rf, target_names=label_encoder_final.classes_, zero_division=0))
```
