```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_raw = pd.read_csv('NVIDIA_STOCK.csv')
correct_headers = ['Date', 'Adj Close', 'Close', 'High', 'Low', 'Open', 'Volume']

df = df_raw.iloc[2:].copy()
df.columns = correct_headers

df['Date'] = pd.to_datetime(df['Date'])

numeric_columns = ['Adj Close', 'Close', 'High', 'Low', 'Open', 'Volume']
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df = df.dropna().sort_values('Date')

plt.figure(figsize=(12, 6))

sns.set_theme(style="whitegrid")

plt.plot(df['Date'], df['Close'], color='#76b900', linewidth=2, label='Closing Price ($)')

plt.title('NVIDIA (NVDA) Stock Price Evolution', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price ($)', fontsize=12)
plt.legend(loc='upper left')

plt.tight_layout()

plt.show()
```
